ARRAY INDEXING AND SLICING IN VECTOR


```python
import numpy as np
```

```python
arr[3]
```

ARRAY INDEXING AND SLICING OF MATRIX

```python
arr = np.arange(1,31).reshape(6,5)
arr
```

```python
arr[1]
```

```python
slice = arr[0:2,1:3]
slice
```

```python
slice = arr[3:,3:]
slice
```

```python
arr[:,2]
```

BOOLEAN INDEXING

```python
bool_index = arr % 2 == 0
bool_index
```

```python
arr = arr[bool_index]
arr
```
