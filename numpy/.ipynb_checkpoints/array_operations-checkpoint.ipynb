ARITHMETIC OPERATION

```python
import numpy as np
```

```python
arr1 = np.array([1,2,3,4,5])
arr2 = np.array([6,7,8,9,10])
```

```python
arr1 + arr2
```

```python
arr1 - arr2
```

```python
arr1 * arr2
```

```python
arr1 / arr2
```

```python
arr1 // arr2
```

```python
arr1 ** arr2
```

DEEP AND SHALLOW COPY


Shallow copy

```python
a = np.arange(1,11)
a
```

```python
b = a[1:5]
b
```

```python
b + 10
```

```python
a
```

Deep Copy

```python
a = [1,2,3,4]
b = a.copy()
```

```python
b[0] = 10
```

```python
b
```

```python
a
```

Advance Array Operations

Stack -> Horizental stack -> Vertical Stack -> Column Stack

```python
a = np.array([1,2,3])
```

```python
b = np.array([4,5,6])
```

Horizental stack makes the array 1D -> 1 Dimensional

```python
np.hstack((a,b))
```

Vertical Stack makes the array 2D -> 2 Dimensional

```python
np.vstack((a,b))
```

columnStack makes the array in column wise like the columns becomes the row 

```python
np.column_stack((a,b))
```
